```python
import sys
from pathlib import Path

import pandas as pd

project_root = Path.cwd().parent
sys.path.append(str(project_root))

from src.data_loader import download_market_data
```

```python
portfolio_assets = [
    "AAPL",  # Apple
    "JPM",   # JPMorgan Chase
    "JNJ",   # Johnson & Johnson
    "EFA",   # Developed Markets Equity ETF
    "LQD",   # Investment Grade Corporate Bond ETF
    "DBC"    # Broad Commodity ETF
]

# Benchmark only
benchmark = "SPY"

# Download all holdings plus benchmark
tickers = portfolio_assets + [benchmark]

print("Portfolio holdings:", portfolio_assets)
print("Benchmark:", benchmark)
print("All download tickers:", tickers)
```

```python
from datetime import date

start_date = "2017-01-01"
end_date = date.today().strftime("%Y-%m-%d")

print(f"Start Date: {start_date}")
print(f"End Date: {end_date}")
```

```python
market_data = download_market_data(
    tickers=tickers,
    start_date=start_date,
    end_date=end_date
)
```

```python
# Inspect downloaded market data
print("Downloaded data shape:", market_data.shape)

print("\nDownloaded tickers:")
print(market_data.columns.get_level_values(0).unique().tolist())

print("\nAvailable data fields:")
print(market_data.columns.get_level_values(1).unique().tolist())

display(market_data.head())
```

```python
# Extract adjusted closing prices
close_prices = market_data.xs(
    "Close",
    level=1,
    axis=1
)

# Reorder columns according to our portfolio configuration
close_prices = close_prices.reindex(columns=tickers)

# Display the first five observations
display(close_prices.head())
```

```python
print("Column order:")
print(close_prices.columns.tolist())

print("\nDataset dimensions:")
print(close_prices.shape)
```

```python
close_prices.shape
```

```python
close_prices.describe()
```

```python
# Dataset quality checks

print("Missing values:")
print(close_prices.isnull().sum())

print("\nDuplicate dates:")
print(close_prices.index.duplicated().sum())

print("\nNon-positive prices:")
print((close_prices <= 0).sum())

print("\nFirst available date:", close_prices.index.min())
print("Last available date:", close_prices.index.max())

print("\nChronologically sorted:",
      close_prices.index.is_monotonic_increasing)
```

```python
from pathlib import Path

processed_path = project_root / "data" / "processed"
processed_path.mkdir(parents=True, exist_ok=True)

close_prices.to_csv(processed_path / "close_prices.csv")

print("Saved successfully!")
```
